## Part 1: OLS Linear Regression

In Financial Econometrics Module 1 Lesson 1 you studied how OLS regression models the relationship between a target stock's excess return and explanatory factors. Following the three-step framework — *identify factors → quantify relationships → forecast* — you will now apply OLS regression to your personally assigned stocks.

Your regression model is:

$$r_{\text{target},t}^{\text{excess}} = \alpha + \beta_1 r_{\text{peer1},t} + \cdots + \beta_4 r_{\text{peer4},t} + \beta_5 r_{\text{DJI},t} + \varepsilon_t$$

> **Reminder:** `y`, `X`, and `X_with_const` are available from the Setup cell.

### FE610-GCA-M1-P1-Q0 Coding Task

Implement `fit_ols_model(y, X)` that fits an OLS model and returns key metrics.

**Inputs:**
- `y` — pandas Series: target stock quarterly excess returns
- `X` — pandas DataFrame: exogenous variables **without** intercept column (peers + DJI, 5 cols)

**Output:** A Python `dict` with exactly these keys:

| Key | Type | Description |
|---|---|---|
| `'r_squared'` | `float` | R², rounded to 6 decimal places |
| `'adj_r_squared'` | `float` | Adjusted R², rounded to 6 decimal places |
| `'coefficients'` | `np.ndarray` | `[intercept, β_peer1, β_peer2, β_peer3, β_peer4, β_DJI]` |

**Requirements:** use `statsmodels.api.OLS`; add intercept with `sm.add_constant(X, prepend=True)` **inside** your function; return `np.array(model.params)` for coefficients.

```python
result = fit_ols_model(y, X)
print(result['r_squared'])      # e.g. 0.312456
print(result['adj_r_squared'])  # e.g. 0.287103
print(result['coefficients'])   # e.g. array([ 0.002, 0.45, -0.12, 0.33, 0.08, 0.61])
```

````python
# FE610-GCA-M1-P1-Q0 Coding Task — Starter Code

random.seed(studentID)

def fit_ols_model(y, X):
    """
    Fit OLS linear regression and return key metrics.

    Parameters
    ----------
    y : pd.Series      — target stock quarterly excess returns
    X : pd.DataFrame   — exogenous variables WITHOUT intercept column

    Returns
    -------
    dict with keys:
        'r_squared'     : float      — R2, rounded to 6 d.p.
        'adj_r_squared' : float      — Adjusted R2, rounded to 6 d.p.
        'coefficients'  : np.ndarray — [intercept, b_peer1, ..., b_peer4, b_DJI]
    """
    # Step 1: Add an intercept term to X so the model includes a constant.
    #   OLS requires a constant column (intercept α); without it the regression
    #   is forced through the origin, which is rarely appropriate.
    #   ⚠ Add the constant INSIDE this function — the grader passes X without one.

    # Step 2: Fit OLS on the augmented design matrix and y.
    #   You will need the model's R², adjusted R², and parameter estimates.

    # Step 3: Return a dict with 'r_squared', 'adj_r_squared', and 'coefficients'.
    #   Round the scalar metrics to 6 decimal places.
    #   Return the full coefficient array as np.ndarray preserving the order
    #   [intercept, peer1, peer2, peer3, peer4, DJI].
    pass


random.seed(studentID)


def fit_ols_model(y, X):
    """Fit OLS linear regression and return key metrics."""
    # Step 1: Add intercept column to X inside the function
    X_const = sm.add_constant(X, prepend=True)

    # Step 2: Fit OLS on the augmented design matrix
    model = sm.OLS(y, X_const).fit()

    # Step 3: Return dictionary with required keys, rounding, and types
    return {
        "r_squared": round(float(model.rsquared), 6),
        "adj_r_squared": round(float(model.rsquared_adj), 6),
        "coefficients": np.array(model.params),
    }
````

````python
# Inspect your result — run after implementing the function above
ols_result = fit_ols_model(y, X)

print(f'R2           : {ols_result["r_squared"]}')
print(f'Adjusted R2  : {ols_result["adj_r_squared"]}')
print(f'Coefficients :')
for name, val in zip(['const'] + peer_tickers + ['DJI'], ols_result['coefficients']):
    print(f'  {name:<12}: {val:.6f}')
````

## Part 2: Correlation Matrix and Multicollinearity

In Module 1 Lesson 2 you studied how to detect multicollinearity using the Pearson correlation matrix and the Variance Inflation Factor (VIF). Multicollinearity occurs when exogenous variables are highly linearly related, which inflates standard errors and makes individual coefficient estimates unreliable.

You will now compute the full correlation matrix for your exogenous variables (peers + DJI) and calculate VIF scores to assess the extent of multicollinearity in your specific stock assignment.

$$\text{VIF}_j = \frac{1}{1 - R^2_j}$$

where $R^2_j$ is the R² from regressing variable $j$ on all other exogenous variables.

> **Reminder:** `df_exog` (peers + DJI, no target stock, no intercept) is available from the Setup cell.

### FE610-GCA-M1-P2-Q0 Coding Task

Implement `compute_correlation_and_vif(df)` that returns the Pearson correlation matrix and VIF scores for all variables in `df`.

**Inputs:**
- `df` — pandas DataFrame: exogenous variables (peers + DJI), **no intercept column**

**Output:** A Python `dict` with exactly these keys:

| Key | Type | Description |
|---|---|---|
| `'correlation_matrix'` | `pd.DataFrame` | Pearson correlation matrix (shape n_vars × n_vars) |
| `'vif_series'` | `pd.Series` | VIF score per variable, indexed by column name |

**Requirements:** use `df.corr()` for the correlation matrix; use `variance_inflation_factor(df.values, i)` from `statsmodels.stats.outliers_influence` for VIF.

```python
result = compute_correlation_and_vif(df_exog)
print(result['correlation_matrix'].round(4))
print(result['vif_series'].round(4))
```

````python
# FE610-GCA-M1-P2-Q0 Coding Task — Starter Code

random.seed(studentID)

def compute_correlation_and_vif(df):
    """
    Compute the Pearson correlation matrix and VIF scores for all variables.

    Parameters
    ----------
    df : pd.DataFrame — exogenous variables (no intercept column)

    Returns
    -------
    dict with keys:
        'correlation_matrix' : pd.DataFrame — Pearson correlation matrix
        'vif_series'         : pd.Series    — VIF score per variable
    """
    # Step 1: Compute the pairwise Pearson correlation matrix for all columns in df.
    #   The result is a symmetric DataFrame with 1.0 on the diagonal.

    # Step 2: Compute VIF for each variable.
    #   VIF_j = 1 / (1 - R2_j) where R2_j comes from regressing column j on all others.
    #   Loop over column indices 0 ... df.shape[1]-1 and apply variance_inflation_factor.
    #   Collect results into a pd.Series indexed by df.columns.
    #   ⚠ Pass df.values (numpy array), not df itself, to variance_inflation_factor.

    # Step 3: Return a dict with 'correlation_matrix' and 'vif_series'.
    pass

random.seed(studentID)


def compute_correlation_and_vif(df):
    """Compute the Pearson correlation matrix and VIF scores for all variables."""
    # Step 1: Compute pairwise Pearson correlation matrix
    correlation_matrix = df.corr()

    # Step 2: Compute VIF for each column using df.values and index by df.columns
    vif_values = [
        variance_inflation_factor(df.values, i) for i in range(df.shape[1])
    ]
    vif_series = pd.Series(vif_values, index=df.columns)

    # Step 3: Return dictionary with exact keys
    return {
        "correlation_matrix": correlation_matrix,
        "vif_series": vif_series,
    }
````

````python
# Inspect your result — run after implementing the function above
corr_result = compute_correlation_and_vif(df_exog)

print('Correlation matrix:')
print(corr_result['correlation_matrix'].round(4))
print()
print('VIF scores:')
print(corr_result['vif_series'].round(4))
````

## Part 3: Principal Component Analysis

In Module 1 Lesson 3 you studied how PCA extracts the dominant common factors driving correlated variable movements. Applied to stock returns, PCA reveals the latent market factors — e.g. a broad equity beta factor, a sector-rotation factor — that explain most of the co-movement across your assigned stocks.

You will apply PCA to your standardised stock returns (target + 4 peers) and summarise how many principal components are required to capture 90% of total variance.

> **Reminder:** `standardized_returns` (zero-mean, unit-variance returns for all 5 assigned tickers) and `all_tickers` are available from the Setup cell.

### FE610-GCA-M1-P3-Q0 Coding Task

Implement `apply_pca(standardized_returns, n_components)` that performs PCA on standardised stock returns.

**Inputs:**
- `standardized_returns` — pandas DataFrame: zero-mean, unit-variance quarterly returns (5 stocks)
- `n_components` — int: number of principal components to extract

**Output:** A Python `dict` with exactly these keys:

| Key | Type | Description |
|---|---|---|
| `'explained_variance_ratio'` | `np.ndarray` | Proportion of variance explained by each component |
| `'cumulative_variance'` | `np.ndarray` | Cumulative proportion from PC1 onwards |
| `'n_components_90'` | `int` | Minimum components needed to reach ≥ 90% cumulative variance |

**Requirements:** use `sklearn.decomposition.PCA`; fit a *separate* unconstrained `PCA()` (all components) to compute `n_components_90`.

```python
result = apply_pca(standardized_returns, n_components=len(all_tickers))
print(result['explained_variance_ratio'])  # e.g. array([0.376, 0.237, 0.171, 0.115, 0.101])
print(result['cumulative_variance'])       # e.g. array([0.376, 0.613, 0.784, 0.899, 1.000])
print(result['n_components_90'])           # e.g. 5
```

````python
# FE610-GCA-M1-P3-Q0 Coding Task — Starter Code

random.seed(studentID)

def apply_pca(standardized_returns, n_components):
    """
    Apply PCA to standardised stock returns.

    Parameters
    ----------
    standardized_returns : pd.DataFrame — zero-mean, unit-variance returns
    n_components         : int          — number of components to extract

    Returns
    -------
    dict with keys:
        'explained_variance_ratio' : np.ndarray — variance ratio per component
        'cumulative_variance'      : np.ndarray — cumulative variance ratio
        'n_components_90'          : int        — components needed for >= 90%
    """
    # Step 1: Fit PCA with the specified n_components.
    #   This gives explained_variance_ratio_ for the requested number of components.

    # Step 2: Fit a second unconstrained PCA (no n_components argument) to obtain
    #   the cumulative variance profile across ALL components.
    #   Use this to find the first position where cumulative variance reaches 0.90.
    #   n_components_90 = index of first True in (cumulative >= 0.90), plus 1.

    # Step 3: Return a dict with 'explained_variance_ratio', 'cumulative_variance',
    #   and 'n_components_90'. The first two come from the Step 1 PCA;
    #   n_components_90 is derived from the Step 2 full PCA.
    pass

random.seed(studentID)


def apply_pca(standardized_returns, n_components):
    """Apply PCA to standardised stock returns."""
    # Step 1: Fit PCA with the specified n_components
    pca = PCA(n_components=n_components)
    pca.fit(standardized_returns)
    explained_variance_ratio = np.array(pca.explained_variance_ratio_)
    cumulative_variance = np.cumsum(explained_variance_ratio)

    # Step 2: Fit an unconstrained PCA() across all components for 90% threshold
    pca_full = PCA()
    pca_full.fit(standardized_returns)
    cum_full = np.cumsum(pca_full.explained_variance_ratio_)
    n_components_90 = int(np.argmax(cum_full >= 0.90)) + 1

    # Step 3: Return dictionary with exact keys and types
    return {
        "explained_variance_ratio": explained_variance_ratio,
        "cumulative_variance": cumulative_variance,
        "n_components_90": n_components_90,
    }
````

## Part 4: Monte Carlo Investment Simulation

In Module 1 Lesson 4 you studied how Monte Carlo simulation models the distribution of investment outcomes under stochastic return dynamics. By drawing many random paths from a return distribution, you can estimate not just the expected outcome but the full range of possibilities — including tail risks.

You will now simulate a monthly investment plan using the return statistics of your target stock, and summarise the resulting distribution of final balances.

> **Reminder:** `target_mean_qtr` and `target_std_qtr` are the quarterly mean and standard deviation of your target stock's excess returns, available from the Setup cell.

### FE610-GCA-M1-P4-Q0 Coding Task

Implement `simulate_investment(P, r_mean, r_std, n_months, n_simulations, seed)` that simulates a monthly investment plan and returns summary statistics.

**Inputs:**
- `P` — float: monthly contribution in dollars
- `r_mean` — float: mean annual return
- `r_std` — float: standard deviation of annual return
- `n_months` — int: investment horizon in months
- `n_simulations` — int: number of Monte Carlo trials
- `seed` — int: random seed for reproducibility

**Output:** A Python `dict` with exactly these keys:

| Key | Type | Description |
|---|---|---|
| `'mean_balance'` | `float` | Mean final balance across simulations, rounded to 4 d.p. |
| `'std_balance'` | `float` | Standard deviation of final balance, rounded to 4 d.p. |
| `'percentile_5'` | `float` | 5th percentile final balance, rounded to 4 d.p. |
| `'percentile_95'` | `float` | 95th percentile final balance, rounded to 4 d.p. |

**Approach:** each simulation draws `n_months` annual returns from N(`r_mean`, `r_std`), converts them to monthly compounding rates, then calculates the final balance as the sum of P × (compounding factor for each monthly contribution). Contribution at month `i` grows for `n_months - i` remaining months.

```python
result = simulate_investment(P=10, r_mean=target_mean_qtr*4, r_std=target_std_qtr*2,
                             n_months=12, n_simulations=10000, seed=studentID)
print(result['mean_balance'])    # e.g. 138.9610
print(result['percentile_5'])    # e.g. 126.0796
```

````python
# FE610-GCA-M1-P4-Q0 Coding Task — Starter Code

random.seed(studentID)

def simulate_investment(P, r_mean, r_std, n_months, n_simulations, seed):
    """
    Monte Carlo simulation of a monthly investment plan.

    Parameters
    ----------
    P             : float — monthly contribution in dollars
    r_mean        : float — mean annual return
    r_std         : float — std dev of annual return
    n_months      : int   — investment period in months
    n_simulations : int   — number of Monte Carlo trials
    seed          : int   — random seed for reproducibility

    Returns
    -------
    dict with keys:
        'mean_balance'   : float — mean final balance, rounded to 4 d.p.
        'std_balance'    : float — std dev of final balance, rounded to 4 d.p.
        'percentile_5'   : float — 5th percentile, rounded to 4 d.p.
        'percentile_95'  : float — 95th percentile, rounded to 4 d.p.
    """
    # Step 1: Seed the random number generator with the provided seed value.
    #   This ensures reproducibility — the same seed always produces the same results.

    # Step 2: Draw n_simulations × n_months annual returns from a Normal distribution
    #   with mean r_mean and standard deviation r_std.
    #   Shape: (n_simulations, n_months).

    # Step 3: Compute the compounding multiplier for each monthly contribution.
    #   Contribution at month i earns (n_months - i) further months of growth.
    #   Monthly return rate = annual_return / 12.
    #   Multiplier for contribution at month i = (1 + rate_i / 12)^(n_months - i).
    #   Build the full multiplier matrix of shape (n_simulations, n_months).

    # Step 4: The final balance of each simulation is P times the sum of all
    #   compounding multipliers across months. Shape: (n_simulations,).

    # Step 5: Return summary statistics rounded to 4 decimal places:
    #   mean, standard deviation, 5th percentile, and 95th percentile of the balances.
    pass

random.seed(studentID)


def simulate_investment(P, r_mean, r_std, n_months, n_simulations, seed):
    """Monte Carlo simulation of a monthly investment plan."""
    # Step 1: Seed numpy's global random generator for reproducibility
    np.random.seed(seed)

    # Step 2: Draw (n_simulations, n_months) annual returns from N(r_mean, r_std)
    annual_returns = np.random.normal(
        loc=r_mean, scale=r_std, size=(n_simulations, n_months)
    )

    # Step 3: Monthly compounding multiplier for contribution at month i (0 <= i < n_months)
    remaining_months = np.arange(n_months, 0, -1)
    multipliers = (1.0 + annual_returns / 12.0) ** remaining_months

    # Step 4: Final balance per simulation: P * sum of multipliers across months
    final_balances = P * np.sum(multipliers, axis=1)

    # Step 5: Summary statistics rounded to 4 decimal places
    return {
        "mean_balance": round(float(np.mean(final_balances)), 4),
        "std_balance": round(float(np.std(final_balances)), 4),
        "percentile_5": round(float(np.percentile(final_balances, 5)), 4),
        "percentile_95": round(float(np.percentile(final_balances, 95)), 4),
    }
````
